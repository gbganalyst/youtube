# Naming rules for variables

It is good practice to give meanigful variable name so that when someone else or you in future can understand what you were trying to accomplish. The best naming convention is to choose a variable name that will tell the reader of the program what the variable represents.

## Rules for naming variables 

- All variables must begin with a letter of the alphabet

- After the initial letter, variable names can also contain underscore (_), and/or a number. No spaces or special characters, however are allowed

- Uppercase strings are different from lowercase string in Python

# How to get information about a function?

If you want to become a good programmer, it is important to read functions documentations. This is like a user guide or manual about a certain function. There are many ways to get help in Python, but the simplest way is to question the function by using question mark. For example, to read the help file or documentation of `print()` function, we shall use `?print`.


```python
?print
```

It prints the values of an object.

```python
print(7)
```

```python
x = 7
```

```python
print(x)
```

Let’s see what `float()` function does by using `?float`

```python
?float
```

It converts a string or number to a floating point number, if possible.

# Warning

Don't use parentheses in the function when seeking for help. For example if you need a help on what int() function does, just type int without parentheses.

```python
?int
```

If you include parentheses, you will have the following:


```python
?int
```

```python
?print
```

---
As you can see, that is not what you are looking for.

```python
help(print)
```
