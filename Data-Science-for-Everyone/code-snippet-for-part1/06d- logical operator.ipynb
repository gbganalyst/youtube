# Logical operators 

## Using the `and` operator

`and` - This operator return a False when it encounters a `False` condition


### Example

Let `mins` represent the number of minutes a money transfer transaction in South Sudan takes to complete.

```python
mins = 3
```

```python
mins < 7
```

```python
mins < 1
```

```python
# Run this cell to see the outcome

mins < 7 and mins < 1
```

Since the two expressions are both False i.e. 

`mins < 7`  is `False`  

`mins < 1` is `False` 

Therefore, `mins < 7 and mins < 1` will be `False`

```python
age = 20
```

```python
age > 17
```

```python
age == 20
```

```python
age > 17 and age == 20
```

## Using the `or` operator

`or` -  This operator returns  `True` when it encounters at least a `True` condition


### Example 

Let `year` represent the number of years Aden a Somalia has been living in Ethipia


```python
year = 12
```

```python
year == 12
```

```python
year < 10
```

```python
# Run this cell to see the outcome

year == 12 or  year < 10
```

Since one expression is True i.e. 

`year == 12 `  is `True`  

`year < 10 ` is `False` 

Therefore, `year == 12 or  year < 10` will be `True`

```python
covid19 = 9
```

```python
covid19 <= 8
```

```python
covid19 >= 19
```

```python
covid19 <= 8 or covid19 >= 19
```

## Using the `not` operator

`not`-  This operator returns `True` if the condition is `False` and `True` if the condition is `False`

### Example 

We define a variable `mins` to represent the number of minutes a money transfer transaction in Sudan takes to be completed.

```python
mins = 10
```

```python
mins > 5
```

```python
not mins > 5
```

As you know, anything that is not `True` is `False`! For example:

```python
not True
```

```python
not False
```
