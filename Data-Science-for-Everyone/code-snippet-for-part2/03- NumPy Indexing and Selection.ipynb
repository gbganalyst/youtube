# Indexing and Slicing NumPy Arrays

Once you have created arrays, the next important skill is learning how to *access* and *work with* specific elements, rows, columns, or blocks inside those arrays. This process is called **indexing** and **slicing**.

NumPy indexing is very similar to what you already know from Python lists, but with powerful extensions that are extremely useful in data science. The same ideas will later appear again when we work with pandas `Series` and `DataFrame` objects.

NumPy uses **zero-based indexing**, so the first element is at index `0`, the second at `1`, and so on.

---

## Indexing and Slicing One-Dimensional Arrays

Let us start with a simple one-dimensional array of ages:

| Positive Index | 0 | 1 | 2 | 3 | 4 | 5 |
|----------------|---|---|---|---|---|---|
| Ages           | 18| 20| 22| 19| 21| 23|
|**Negative Index** | **-6**| **-5**| **-4**| **-3**| **-2**|**-1**|

```python
import numpy as np

ages = np.array([18, 20, 22, 19, 21, 23])
ages
```

#### Accessing single elements

```python
ages[0]      # first element (18)
```

```python
ages[-1]     # last element (23)
```

```python
ages[2]      # third element (22)
```

#### Basic slicing with start, stop, and step

```python
ages[1:4]    # elements at index 1, 2, 3 -> [20, 22, 19]
```

```python
ages[:3]     # first three elements -> [18, 20, 22]
```

```python
ages[3:]     # from index 3 to the end -> [19, 21, 23]
```

```python
ages[::2]    # every second element -> [18, 22, 21]
```

```python
ages[::-1]   # reverse the array -> [23, 21, 19, 22, 20, 18]
```

#### Modifying elements through indexing

Indexing also allows you to change values in place. This is a common operation in data cleaning and preprocessing.

```python
ages[0] = 19          # correct a data entry
```

```python
ages[2:4] = [21, 20]  # update a slice of the array
```

```python
ages
```

#### Boolean indexing (filtering values)

Boolean indexing is particularly important in data science. You create a logical condition and use it to filter the array.

```python
ages = np.array([18, 20, 22, 19, 21, 23])

# create a boolean mask
mask = ages >= 21
mask
```

```python
# use the mask to filter the array
adult_ages = ages[mask]
adult_ages
```

```python
# you can also write it in one line
ages[ages < 20]   # all ages less than 20
```

This style of filtering is exactly the kind of operation we will later perform on pandas `Series` and `DataFrame` columns when we subset rows based on conditions.

## Indexing and Slicing Two-Dimensional Arrays

Two-dimensional arrays represent tables or matrices. Think of them as data with rows and columns, similar to a spreadsheet or a pandas `DataFrame`.

```python
scores = np.array([[65, 70, 80],
                   [75, 85, 90],
                   [55, 60, 65]])
scores
```

Conceptually,

<td width="50%">

**Columns:**
* Column 0: `[65, 75, 55]` 
* Column 1: `[70, 85, 60]`
* Column 2: `[80, 90, 65]`

</table>

#### Accessing single elements

To access a single value, you provide `[row_index, column_index]`.

```python
scores[0, 0]   # first row, first column -> 65
```

```python
scores[1, 2]   # second row, third column -> 90
```

```python
scores[2, 1]   # third row, second column -> 60
```

#### Accessing entire rows or columns

```python
scores[0]        # entire first row -> [65, 70, 80]
```

```python
scores[1]        # second row -> [75, 85, 90]
```

```python
scores[:, 0]     # entire first column -> [65, 75, 55]
```

```python
scores[:, 2]     # entire third column -> [80, 90, 65]
```

#### Slicing sub-matrices

You can slice out rectangular blocks of the table.

```python
scores[:, 1:3]   # all rows, columns 1 and 2
```

```python
scores[0:2, :]   # first two rows, all columns
```

```python
scores[1:, 1:]   # from row 1 and column 1 to the end
```

#### Modifying values in a matrix

```python
# increase all scores in the first row by 5
scores[0, :] = scores[0, :] + 5
```

```python
# set the entire third column to 100
scores[:, 2] = 100

scores
```

Operations like “update all scores in a column” or “adjust scores for some rows” are very common in data science workflows.

###  More Examples: Practical Data Science Operations

Below are a few more examples that mirror real data science tasks.

#### Example 1: Normalizing exam scores

Suppose the scores for the second test are stored in the second column. We may want to scale them to a 0–1 range for modeling.

```python
scores = np.array([[65, 70, 80],
                   [75, 85, 90],
                   [55, 60, 65]])

second_test = scores[:, 1]          # extract the second column
min_val = second_test.min()
max_val = second_test.max()

scaled_second_test = (second_test - min_val) / (max_val - min_val)
scaled_second_test
```

#### Example 2: Selecting students who scored above a threshold

```python
total_scores = scores.sum(axis=1)   # sum across columns for each row
total_scores
```

```python
# create a mask for students with total score at least 200
high_performers_mask = total_scores >= 200
high_performers_mask
```

```python
# select only rows for high performing students
high_performers = scores[high_performers_mask, :]
high_performers
```

This is directly comparable to selecting rows in a pandas `DataFrame` where a total score column exceeds a threshold.

#### Example 3: Selecting every other row and column

```python
# every other row, every other column
subsample = scores[::2, ::2]
subsample
```

Subsampling like this is useful for quick checks, exploratory analysis, or when you are working with very large datasets.

### 6.4 How This Connects to pandas

The ideas you see here are the foundation for how we work with data in pandas:

* Selecting columns in a `DataFrame` feels similar to slicing columns in a NumPy array.
* Selecting specific rows and columns using `.iloc` in pandas mirrors the `[row_index, column_index]` style in NumPy.
* Filtering rows using conditions in pandas uses the same idea as boolean indexing in NumPy.

For example, the following operations:

```python
scores[:, 0]      # all rows, first column
```
