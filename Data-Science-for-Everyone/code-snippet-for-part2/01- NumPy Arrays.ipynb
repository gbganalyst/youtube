## 3. Python Lists vs NumPy Arrays

We start with a familiar structure: the Python list.

Then we compare it with NumPy arrays and see why NumPy is the preferred choice in data science.

### 3.1 Python list



````python
heights_list = [160, 165, 170, 172, 168]
type(heights_list)
````

* A Python list can hold different types, is flexible, but arithmetic is not vectorised.
* For example, trying `heights_list * 2` repeats the list instead of doubling values.

````python
len(heights_list * 2)
````

````python
heights_list + 1
````

### 3.2 NumPy array

````python
heights = np.array([160, 165, 170, 172, 168])
type(heights)
````


Now arithmetic is vectorised:

````python
heights * 2
````

````python
heights + 1
````

````python
np.sqrt(heights)
````

This behaviour is essential for data science, because it allows us to work with whole columns of data efficiently, just like in pandas.

### Normal (Gaussian) Distribution with `random.normal`

$X \sim N(\mu, \sigma)$ 

**Example 1**:

Generate 100 random numbers from a normal distribution with mean = 60 and standard deviation = 10.

````python
data = np.random.normal(loc=60, scale=10, size=100) # mean=60, std=10, size=100

plt.hist(data)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Normally Distributed Data")
plt.show()
````

**Conducting Statistical Hypothesis Tests**

$H_0$: The data follows a normal distribution.

$H_1$: The data does not follow a normal distribution.

````python
from scipy.stats import shapiro
````

````python
# Shapiro-Wilk test
stat, p_value = shapiro(data)
print("Shapiro-Wilk Test Statistic:", stat)
print("p-value:", p_value)

# Interpretation
alpha = 0.05
if p_value > alpha:
    print("Fail to reject the null hypothesis. Data appear normally distributed.")
else:
    print("Reject the null hypothesis. Data do not appear normally distributed.")
````

### Reshaping NumPy Arrays with `reshape`

Sometimes you already have an array, but you want to **change its shape** without changing the underlying data. The NumPy `reshape()` method allows you to do exactly that.

At its core:

> `reshape` changes the **view** (shape) of the data, not the data itself.

#### Basic idea

````python
import numpy as np

a = np.array([65, 70, 80, 75, 85, 90])
a.shape 
````

You can reshape this one dimensional array into a matrix with 2 rows and 3 columns:

````python
b = a.reshape(2, 3)
print(b)
````

* Original shape: `(6,)`
* New shape: `(2, 3)`
* The 6 values are the same. Only the arrangement has changed.

#### Rule: total number of elements must match

The product of the new dimensions must equal the number of elements in the array.

````python
np.arange(12).reshape(3, 4)   # 3 × 4 = 12  → valid
````

````python
np.arange(12).reshape(5, 3)   # 5 × 3 = 15  → error
````

If the numbers do not match, NumPy raises an error.
